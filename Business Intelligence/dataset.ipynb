```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('health_fitness_dataset.csv')  # Adjust filename as needed

# --- 1. Workout Trends Across Demographics ---
def plot_workout_trends():
    plt.figure(figsize=(12, 6))
    sns.countplot(data=df, x='activity_type', hue='age')  # Modify column names accordingly
    plt.title('Workout Trends Across Ages')
    plt.xticks(rotation=45)
    plt.show()

# --- 2. Most Effective Activity Types (Calories Burned) ---
def plot_calories_burned():
    plt.figure(figsize=(12, 6))
    sns.boxplot(data=df, x='activity_type', y='calories_burned')
    plt.title('Calories Burned by Activity Type')
    plt.xticks(rotation=45)
    plt.show()

# --- 3. Impact of Smoking on Health Indicators ---
def plot_smoking_effects():
    metrics = ['stress_level', 'hours_sleep', 'duration_minutes', 'weight_kg']  # Adjust as per dataset
    df_melted = df.melt(id_vars=['smoking_status'], value_vars=metrics, var_name='Health Indicator', value_name='Value')
    
    plt.figure(figsize=(12, 6))
    sns.barplot(data=df_melted, x='Health Indicator', y='Value', hue='smoking_status')
    plt.title('Effects of Smoking on Health Indicators')
    plt.xticks(rotation=45)
    plt.show()

# --- 4. Demographics & Fitness Performance ---
def plot_demographics_fitness():
    plt.figure(figsize=(12, 6))
    sns.scatterplot(data=df, x='age', y='daily_steps', hue='bmi', palette='coolwarm')
    plt.title('Step Count vs Age (Colored by BMI)')
    plt.show()



# Run the plots
plot_workout_trends()
plot_calories_burned()
plot_smoking_effects()
plot_demographics_fitness()

```

```python
# --- 5. BMI vs. Physical Fitness ---
def plot_bmi_vs_fitness():
    plt.figure(figsize=(12, 6))
    sns.regplot(data=df, x='bmi', y='calories_burned')
    plt.title('BMI vs Calories Burned')
    plt.show()

# --- 6. Hydration vs. Heart Rate ---
def plot_hydration_vs_hr():
    plt.figure(figsize=(12, 6))
    sns.scatterplot(data=df, x='hydration_level', y='avg_heart_rate')
    plt.title('Hydration Level vs Average Heart Rate')
    plt.show()
    
plot_bmi_vs_fitness()
plot_hydration_vs_hr()
```
